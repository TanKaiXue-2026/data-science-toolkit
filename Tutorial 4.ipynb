```python
1. Is this data in tidy format?
No, this dataset is in a untidy format.
```

```python
2. If not, explain which tidy-data rule is violated.
The dataset does not follow the first fundamental rule of tidy data: "Each variable must have its own column."
```

```python
import pandas as pd

data1 = {
    "Student": ["Ali", "Sara", "John"],
    "Math": [75, 88, 65],
    "Science": [80, 91, 70],
    "English": [72, 85, 68]
}

a = pd.DataFrame(data1)
a_tidy = a.melt(id_vars=["Student"], var_name="Subject", value_name="Score")

print(a_tidy)
```

```python
1. What is the tidy-data problem with this dataset?
The column names Q1_Sales, Q2_Sales, Q3_Sales, and Q4_Sales contain values of a variable (Quarter) rather than distinct variable names.
A tidy structure requires a single Quarter column to store the quarter names and a Sales column for the numeric values.
```

```python
import pandas as pd

# Define dataset
data2 = {
    "Country": ["Malaysia", "Thailand", "Indonesia"],
    "Q1_Sales": [12000, 10000, 14000],
    "Q2_Sales": [15000, 11000, 12500],
    "Q3_Sales": [14000, 13000, 15000],
    "Q4_Sales": [18000, 16000, 17000]
}

df2 = pd.DataFrame(data2)

# Reshape to tidy format
df2_tidy = df2.melt(id_vars=["Country"], var_name="Quarter", value_name="Sales")

# Clean up Quarter column (remove '_Sales' suffix)
df2_tidy["Quarter"] = df2_tidy["Quarter"].str.replace("_Sales", "")

print(df2_tidy)
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Load dataset
data = {
    "Student": ["A", "B", "C", "D", "E", "F"],
    "Study_Hours": [2, 5, 3, 7, 4, 8],
    "Score": [55, 72, 60, 85, 68, 90]
}
df = pd.DataFrame(data)

# 1. Average study time
avg_study = df["Study_Hours"].mean()

# 2. Average score
avg_score = df["Score"].mean()

# 3. Student with highest score
top_score_student = df.loc[df["Score"].idxmax(), "Student"]

# 4. Student who studied the most
top_study_student = df.loc[df["Study_Hours"].idxmax(), "Student"]

print(f"1. Average Study Time : {avg_study:.2f} hours")
print(f"2. Average Score      : {avg_score:.2f}")
print(f"3. Highest Score      : Student {top_score_student}")
print(f"4. Most Study Hours   : Student {top_study_student}")
```

```python
5. Yes. Studying more hours is strongly associated with higher scores. The dataset shows a consistent positive trend, where each increase in study hours corresponds to a higher test score.
```

```python
# 6. Scatter plot script
plt.figure(figsize=(7, 5))
plt.scatter(df["Study_Hours"], df["Score"], color="blue", s=70, edgecolor="black")

# Annotate student labels next to data points
for idx, row in df.iterrows():
    plt.annotate(
        f"Student {row['Student']}",
        (row["Study_Hours"], row["Score"]),
        textcoords="offset points",
        xytext=(5, 5)
    )

plt.title("Study Hours vs Score", fontsize=14)
plt.xlabel("Study Hours", fontsize=12)
plt.ylabel("Score", fontsize=12)
plt.grid(True, linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()
```
